```python
# Read the saved HTML file
with open('candidates_2082.html','r',encoding='utf-8') as file:
    html = file.read()
```

```python
# Import needed libraries
from bs4 import BeautifulSoup
import numpy as np 
import pandas as pd 
```

```python
# Parse the HTML
soup = BeautifulSoup(html, 'lxml')
```

```python
# Find all candidate cards
containers = soup.find_all('div', {'class':'p-4 flex flex-col gap-3 flex-1'})
```

```python
len(containers)
```

```python
# Create columns
candidate = []
party = []
constituency = []
votes = []
result = []
sex = []
province = []
district = []

for i in containers:
    try:
        candidate.append(i.find('div', {'class':'font-bold text-sm text-slate-900 dark:text-slate-100 leading-tight truncate'}).text)   
    except:
        candidate.append(np.nan)
    try:
        party.append(i.find('span', {'class':'text-[11px] font-medium text-slate-700 dark:text-slate-300 truncate'}).text)
    except:
        price.append(np.nan)
    try:
        constituency.append(i.find('div', {'class':'text-[11px] text-slate-500 dark:text-slate-400 truncate'}).text)
    except:
        constituency.append(np.nan)
    try:
        votes.append(i.find('span', {'class':'text-sm font-extrabold tabular-nums text-slate-800 dark:text-slate-100'}).text)
    except:
        votes.append(np.nan)
    try:
        result.append(i.find('div', {'class':'flex items-center gap-2 mt-auto pt-1'}).find_all('span')[1].text)
    except:
        result.append(np.nan)
    try:
        sex.append(i.find('span', {'class':'text-[10px] text-slate-400'}).text)
    except:
        sex.append(np.nan)
    try:
        province.append(i.find('div', {'class':'flex flex-wrap gap-1'}).find('span').text)
    except:
        province.append(np.nan)
    try:
        district.append(i.find('span', {'class':'text-[10px] font-semibold px-2 py-0.5 rounded-full bg-slate-100 dark:bg-slate-800 text-slate-600 dark:text-slate-400'}).text)
    except:
        district.append(np.nan)
    
```

```python
# Make a dataframe
df = pd.DataFrame({
    'candidate':candidate,
    'party':party,
    'constituency':constituency,
    'votes':votes,
    'result':result,
    'sex':sex,
    'province':province,
    'district':district,
})
```
