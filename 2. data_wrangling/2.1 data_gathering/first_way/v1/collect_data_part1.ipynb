```python
# ---- import libraries ----
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options

from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

import time
```

```python
# ---- browser settings ----
options = Options()
options.add_argument("--disable-blink-features=AutomationControlled")

driver = webdriver.Chrome(
    service=Service(r"C:\Users\Rockit\Desktop\chromedriver.exe"),
    options=options
)

wait = WebDriverWait(driver, 10)
driver.get("https://nepalvotes.live/candidates")
time.sleep(5)
```

```python
all_html = ""

# Loop through pages → 142
for page in range(1, 143):

    print(f"Saving page {page}")

    all_html += f"\n<!-- PAGE {page} START -->\n"
    all_html += driver.page_source
    all_html += f"\n<!-- PAGE {page} END -->\n"

    try:
        next_btn = wait.until(
            EC.element_to_be_clickable(
                (By.XPATH, "//div[contains(@class,'flex')]//button[normalize-space()='→']")
            )
        )

        old_html = driver.page_source

        try:
            next_btn.click()

        except:
            driver.execute_script("arguments[0].click();", next_btn)

        wait.until(lambda d: d.page_source != old_html)
        time.sleep(1)

    except:
        print("No more pages found")
        break
```

```python
# ---- save html ----
html = f"""
<!DOCTYPE html>

<html>

<head>
    <meta charset="utf-8">
    <title>Nepal Votes All Pages</title>
</head>

<body>

{all_html}

</body>

</html>
"""

with open("candidates_2082.html", "w", encoding="utf-8") as file:
    file.write(html)

print("HTML saved successfully")
driver.quit()
```
