```python
candidate = []
party = []
votes = []
province = []
district = []
constituency = []
result = []
sex = []
cid = []

for i in containers:
    try:
        candidate.append(i.find('h1', {'class':'text-2xl sm:text-3xl font-bold text-white leading-tight'}).text)   
    except:
        candidate.append(np.nan)
    try:
        party.append(i.find('span', {'class':'text-sm font-semibold text-white/80'}).text)
    except:
        party.append(np.nan)
    try:
        votes.append(i.find('div', {'class':'text-3xl font-bold tabular-nums text-white leading-none'}).text)
    except:
        votes.append(np.nan)
    try:
        province.append(i.find('span', {'class':'text-[11px] font-semibold px-2.5 py-0.5 rounded-full bg-sky-100 text-sky-800 dark:bg-sky-900/40 dark:text-sky-300'}).text)
    except:
        province.append(np.nan)
    try:
        district.append(i.find('span', {'class':'text-[11px] font-semibold px-2.5 py-0.5 rounded-full bg-white/10 text-white/70'}).text)
    except:
        district.append(np.nan)
    try:
        constituency.append(i.find('div', {'class':'flex flex-wrap gap-2'}).find_all('span')[2].text)
    except:
        constituency.append(np.nan)
    try:
        result.append(i.find('div', {'class':'flex items-center gap-2 flex-wrap'}).find_all('span')[0].text)
    except:
        result.append(np.nan)
    try:
        sex.append(i.find('div', {'class':'flex items-center gap-2 flex-wrap'}).find_all('span')[1].text)
    except:
        sex.append(np.nan)
    try:
        cid.append(i.find('div', {'class':'flex items-center gap-2 flex-wrap'}).find_all('span')[2].text)
    except:
        cid.append(np.nan)
    
```

```python
len(candidate)
```

```python
len(cid)
```

```python
df = pd.DataFrame({
    'candidate':candidate,
    'party':party,
    'votes':votes,
    'province':province,
    'district':district,
    'constituency':constituency,
    'result':result,
    'sex':sex,
    'cid':cid
})
```
