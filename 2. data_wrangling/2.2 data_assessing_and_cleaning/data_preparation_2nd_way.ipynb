#### **batch processing (translation)**

```python
nep_to_eng = GoogleTranslator(source='ne', target='en')

# 1. Get unique, non-null address values to save time and API quota
unique_addresses = df1['address'].dropna().unique().tolist()

# 2. Translate in batches of 50 to avoid request payload limits
batch_size = 50
translated_dict = {}

for i in range(0, len(unique_addresses), batch_size):
    batch = unique_addresses[i:i + batch_size]
    try:
        # Translate the whole list at once
        translations = nep_to_eng.translate_batch(batch)
        # Map original Nepali word to English translation
        for nep, eng in zip(batch, translations):
            translated_dict[nep] = eng
    except Exception as e:
        print(f"Error at batch {i}: {e}")
        # Fallback: if batch fails, keep original Nepali text for this batch
        for nep in batch:
            translated_dict[nep] = nep
```

- *This step takes about **21 minutes** to run. To save time, I did not perform it in the previous steps and executed it only when necessary.*

```python
# 3. Map the translations back to your original series
df1['address'] = df1['address'].map(translated_dict)
```
