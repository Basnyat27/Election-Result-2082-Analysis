```python
temp_df = df.groupby(['constituency', 'party'], as_index=False)['votes'].sum()

# Identify top 10 constituencies by total votes
top_10_const = (
    df.groupby('constituency')['votes'].sum().nlargest(10).index
)

# Replace constituencies not in top 10 with 'Others'
temp_df['constituency_grouped'] = temp_df['constituency'].apply(
    lambda x: x if x in top_10_const else 'Others'
)

# Aggregate to find the average votes for the grouped categories
plot_df = (
    temp_df.groupby(['constituency_grouped', 'party'], as_index=False)[
        'votes'
    ]
    .mean()
    .rename(columns={'constituency_grouped': 'constituency'})
)

fig = px.bar(
    plot_df,
    x='constituency',
    y='votes',
    color='party',
    labels={'votes': 'avg votes'},
    title='Top 10 Constituencies vs Others (Bar Chart)',
)
fig.show()
```

```python
# 1. Find the top 10 constituencies and top 5 parties globally
top_10_const = df.groupby('constituency')['votes'].sum().nlargest(10).index
top_5_parties = df.groupby('party')['votes'].sum().nlargest(5).index

# 2. Sum votes by constituency and party to avoid duplicate rows
temp_df = df.groupby(['constituency', 'party'], as_index=False)['votes'].sum()

# 3. Mask values outside the top groups as 'Others'
temp_df['const_grouped'] = temp_df['constituency'].apply(lambda x: x if x in top_10_const else 'Others')
temp_df['party_grouped'] = temp_df['party'].apply(lambda x: x if x in top_5_parties else 'Others')

# 4. Aggregate by the new grouped columns
plot_df = temp_df.groupby(['const_grouped', 'party_grouped'], as_index=False)['votes'].mean()

# 5. Rename columns for clean plot labels
plot_df = plot_df.rename(columns={'const_grouped': 'constituency', 'party_grouped': 'party'})

# 6. Plot the dynamic stacked bar chart
fig = px.bar(
    plot_df,
    x='constituency',
    y='votes',
    color='party',
    labels={'votes': 'avg votes'},
    title='Top 10 Constituencies vs Others by Top 5 Parties vs Others',
    category_orders={
        'constituency': list(top_10_const) + ['Others'],
        'party': list(top_5_parties) + ['Others']
    }
)
fig.show()
```
